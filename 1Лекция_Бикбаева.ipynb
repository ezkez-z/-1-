```python
#1
n = int(input("Количество человек: "))
s = 0

for i in range(n):
    age = int(input("Введите возраст: "))
    s += age

print("Средний возраст:", s / n)

#2
color = input("Введите код цвета: ").lower()

r = int(color[0:2], 16)
g = int(color[2:4], 16)
b = int(color[4:6], 16)

if color == "000000":
    print("Черный")
elif color == "ffffff":
    print("Белый")
elif r == g == b:
    print("Серый")
elif r > g and r > b:
    print("Красный")
elif g > r and g > b:
    print("Зеленый")
elif b > r and b > g:
    print("Синий")
else:
    print("Смешанный цвет")
```
