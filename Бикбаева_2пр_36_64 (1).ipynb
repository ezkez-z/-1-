```python
#1
employed=input('Трудоустроен?(true или false):')
vacation=input('В отпуске?(true или false):')
if employed==True:
    if vacation==False:
        print(employed)
    else:
        print(False)

else:
    print(employed)

#2
operation=str(input('Операция(+,-,*,/): '))
value_1=int(input('Число1: '))
value_2=int(input('Число2: '))
if operation=='+':
    print(value_1+value_2)
elif operation=='-':
    print(value_1-value_2)
elif operation=='*':
    print(value_1*value_2)
elif operation=='/':
    print(value_1/value_2)

#3
a=int(input('1 число: '))
b=int(input('2 число: '))
summ=0
for x in range(a,b+1):
    summ+=x
print(summ)

#4
month=int(input('Номер месяца: '))
if month in [12,1,2]:
    print('Зима')
    if month in [1,2]:
        print('1 квартал')
    else: print('4 квартал')
elif month in [3,4,5]:
    print('Весна')
    if month in [4,5]:
        print('2 квартал')
    else: print('1 квартал')
elif month in [6,7,8]:
    print('Лето')
    if month in [7,8]:
        print('3 квартал')
    else: print('2 квартал')
elif month in [9,10,11]:
    print('Осень')
    if month in [10,11]:
        print('4 квартал')
    else: print('3 квартал')

#5
summ=0
count=0
while True:
    x=int(input('Оценка(Для отмены введите -1):'))
    if x==-1:
        break
    count+=1
    summ+=x

if count>0:
    sr=summ/count
    print(sr)
if sr>=90: print('A')
elif 80<=sr: print('B')
elif 70<=sr: print('C')
elif 60<=sr: print('D')
else: print('F')

```
