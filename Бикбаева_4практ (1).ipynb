```python
#1.1
def calculate(operation, value_1, value_2):
    if operation == '+':
        return value_1 + value_2
    elif operation == '-':
        return value_1 - value_2
    elif operation == '*':
        return value_1 * value_2
    elif operation == '/':
        return value_1 / value_2
    else:
        print('Неизвестная операция')

def main():
    operation = input('Операция: ')
    value_1 = int(input('Число1: '))
    value_2 = int(input('Число2: '))

    result = calculate(operation, value_1, value_2)
    print(result)

main()

#1.2
def get_season(month):
    if month in [12, 1, 2]:
        return 'Зима'
    elif month in [3, 4, 5]:
        return 'Весна'
    elif month in [6, 7, 8]:
        return 'Лето'
    elif month in [9, 10, 11]:
        return 'Осень'
    else:
        return 'Некорректный номер месяца'

def get_quarter(month):
    if month in [1, 2, 3]:
        return '1 квартал'
    elif month in [4, 5, 6]:
        return '2 квартал'
    elif month in [7, 8, 9]:
        return '3 квартал'
    elif month in [10, 11, 12]:
        return '4 квартал'
    else:
        return 'Некорректный номер месяца'

try:
    month_input = int(input('Введите номер месяца (1-12): '))

    if 1 <= month_input <= 12:
        print(get_season(month_input))
        print(get_quarter(month_input))
    else:
        print('Ошибка: введите число от 1 до 12.')
except ValueError:
    print('Ошибка: введите целое число.')

#2
def count_chars(text, chars):
    counts = {}
    for ch in chars:
        counts[ch] = text.count(ch)
    return counts


text = 'агту вшн лучший!'
chars = 'ашу'

result = count_chars(text, chars)

for ch, cnt in result.items():
    print(ch, ':', cnt)

# 3
def count_m_to_z(text):
    return sum(1 for ch in text if 'm' <= ch <= 'z')

def print_m_to_z(text):
    total_count = count_m_to_z(text)
    print("Количество символов из диапазона от m до z:", total_count)

text = input("Введите строку (только a-z): ").strip()
print_m_to_z(text)


```
