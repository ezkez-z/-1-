```python
import json

try:
    with open('students.json', 'r', encoding='utf-8') as f:
        students = json.load(f)

    if not isinstance(students, list):
        raise TypeError("Данные в файле должны быть списком.")

    student_reports = []
    all_grades = []

    for item in students:
        if not isinstance(item, dict):
            print("Ошибка: Элемент файла не является словарем.")
            continue

        if 'name' not in item or 'grades' not in item:
            print(f"Ошибка: В объекте {item} отсутствуют ключи 'name' или 'grades'.")
            continue

        name = item['name']
        grades = item['grades']

        if not isinstance(name, str) or not isinstance(grades, list):
            print(f"Ошибка: Неверный тип данных для имени или оценок ({name}).")
            continue

        if not grades or not all(isinstance(g, (int, float)) for g in grades):
            print(f"Ошибка: Некорректный список оценок у студента {name}.")
            continue

        avg_student = sum(grades) / len(grades)
        student_reports.append({
            'name': name,
            'average_grade': round(avg_student, 2)
        })

        all_grades.extend(grades)

    if all_grades:
        total_average = round(sum(all_grades) / len(all_grades), 2)
    else:
        total_average = 0

    report = {
        'students': student_reports,
        'total_average': total_average
    }

    with open('report.json', 'w', encoding='utf-8') as f:
        json.dump(report, f, ensure_ascii=False, indent=4)

    print("Отчет успешно создан и сохранен в файл report.json!")

except FileNotFoundError:
    print("Ошибка: Файл 'students.json' не найден!")
except json.JSONDecodeError:
    print("Ошибка: Файл 'students.json' содержит некорректный формат JSON!")
except Exception as e:
    print(f"Произошла непредвиденная ошибка: {e}")
```
