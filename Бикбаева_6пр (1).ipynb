```python
class Student:
    def __init__(self, name, group, grades, tel_number):
        self.name = name
        self.group = group
        self.grades = grades
        self.__tel_number = tel_number

    def add_grade(self, grade):
        self.grades.append(grade)
        print(f"Студенту {self.name} добавлена оценка: {grade}")

    def get_average(self):
        if len(self.grades) == 0:
            return 0
        return sum(self.grades) / len(self.grades)

    def info(self):
        avg = self.get_average()
        print(f"Имя: {self.name}")
        print(f"Группа: {self.group}")
        print(f"Оценки: {self.grades}")
        print(f"Средний балл: {avg:.2f}")
        print(f"Телефон: {self.__tel_number}")
        print("-" * 30)



student1 = Student("Анна", "36-64", [5, 4, 5, 4], "+79991112233")
student2 = Student("Иван", "26-42", [3, 4, 3, 5], "+79992223344")
student3 = Student("Сергей", "36-63", [4, 5, 5, 5], "+79993334455")


print("=== Информация о студентах ===")
student1.info()
student2.info()
student3.info()

print("=== Добавление оценок ===")
student1.add_grade(5)
student2.add_grade(4)
print()

print("=== Обновленная информация ===")
student1.info()
student2.info()
```
